Vamos criar um aplicativo simples de Controle de Inventário que utilize Python e SQL para armazenar, gerenciar e consultar produtos.

Objetivo do Aplicativo:
Desenvolver um aplicativo simples que permita ao usuário cadastrar produtos, atualizar quantidades, visualizar o estoque e gerar um relatório do valor total do inventário.

Tecnologias Usadas:
Python (programação orientada a objetos, manipulação de dados)
SQLite (banco de dados SQL)
Flask (para interface web, se necessário)
SQLAlchemy (ORM para integração com o banco de dados)

Passo 1: Estrutura do Banco de Dados
O banco de dados terá uma tabela produtos com as seguintes colunas:

id (Chave Primária, Integer)
nome (Texto)
quantidade (Inteiro)
preço (Flutuante)

Passo 2: Modelagem em Python com SQLAlchemy
from sqlalchemy import create_engine, Column, Integer, String, Float
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy.orm import sessionmaker

Base = declarative_base()

class Produto(Base):
    __tablename__ = 'produtos'

    id = Column(Integer, primary_key=True)
    nome = Column(String)
    quantidade = Column(Integer)
    preco = Column(Float)

    def __init__(self, nome, quantidade, preco):
        self.nome = nome
        self.quantidade = quantidade
        self.preco = preco

engine = create_engine('sqlite:///estoque.db')
Base.metadata.create_all(engine)

Session = sessionmaker(bind=engine)
session = Session()
